

# 🚀 Setting Up Your AP CSP Working Environment (Ubuntu 24.04.1 LTS)

Welcome! This guide will walk you through setting up your development environment for AP Computer Science Principles on Ubuntu 24.04.1 LTS, from installing the right tools to deploying your website.

## 1. Update Your System 🔄

Open your terminal and run:
```bash
sudo apt update && sudo apt upgrade -y
```

## 2. Install Essential Tools: VS Code and Git 🛠️

Install Git:
```bash
sudo apt install git -y
```

Install VS Code:
- Download the .deb package from the [VS Code website](https://code.visualstudio.com/).
- Install it using:
```bash
sudo apt install ./<path-to-downloaded-file>.deb
```

## 3. Configure Git ⚙️

Set your Git username and email:
```bash
git config --global user.name "Your Name"
git config --global user.email "youremail@example.com"
```

## 4. Clone the Project Repository 🧬

Navigate to your workspace and clone the repository:
```bash
git clone <repository-url>
cd <repository-folder>
```

## 5. Install Python 3 and venv 🐍

Install Python 3 and venv if not already installed:
```bash
sudo apt install python3 python3-venv python3-pip -y
```

## 6. Create and Activate a Virtual Environment 🌱

Create a virtual environment named `venv`:
```bash
python3 -m venv venv
```
Activate the virtual environment:
```bash
source venv/bin/activate
```
(You should see `(venv)` at the start of your terminal prompt.)

## 7. Install Project Dependencies 📦

With the virtual environment activated, install all required Python packages:
```bash
pip install -r requirements.txt
```

## 8. Install VS Code Extensions (Recommended) 🧩

- Python extension for VS Code
- Jupyter extension for running notebooks
- GitLens for Git integration

## 9. Build and Run the Project 🏗️

Run the following command to build and start your project:
```bash
make
```

## 10. Deploy Your Website 🚢

- Follow your project's deployment instructions. This may involve pushing to GitHub Pages, deploying to a cloud provider, or another method.
- For GitHub Pages:
  - Commit and push your changes to the `main` branch.
  - Your site will be automatically deployed if configured correctly.

## 11. Troubleshooting Tips 🛟

- If you encounter errors, check the terminal output for details.
- Make sure all dependencies are installed and up to date.
- Ensure your virtual environment is activated when running Python commands.
- Restart VS Code or your terminal if you run into issues.
- Ask for help if you get stuck!

---
With these steps, you should have a fully working development environment on Ubuntu 24.04.1 LTS and be ready to start coding and deploying your AP CSP projects!
